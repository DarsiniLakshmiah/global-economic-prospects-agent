```python
# Global Economic Prospects Intelligence Agent
# Notebook: 00_environment_check
#
# Purpose:
# Validate the Databricks environment before creating project resources.
#
# This notebook does NOT create or modify project resources.

PROJECT_NAME = "Global Economic Prospects Intelligence Agent"
PROJECT_CATALOG = "worldbank_ai"

print(f"Project: {PROJECT_NAME}")
print(f"Target catalog: {PROJECT_CATALOG}")
```

```python
print("Running basic Spark validation...")

test_df = spark.range(5)

display(test_df)

row_count = test_df.count()

assert row_count == 5, "Spark validation failed."

print("Spark session is working.")
```

```python

import os

print("Environment information")
print("-" * 60)

print("Spark version:", spark.version)

runtime_version = os.environ.get(
    "DATABRICKS_RUNTIME_VERSION",
    "Not exposed through environment variable"
)

print("Databricks Runtime:", runtime_version)
```

```python
# Current identity
identity_df = spark.sql("""
SELECT
    current_user() AS current_user,
    current_catalog() AS current_catalog,
    current_schema() AS current_schema
""")

display(identity_df)
```

```python
print("Catalogs visible to the current identity:")

catalogs_df = spark.sql("SHOW CATALOGS")

display(catalogs_df)
```

```python
# Check whether our catalog already exists
catalog_rows = spark.sql("SHOW CATALOGS").collect()

available_catalogs = {
    row["catalog"]
    for row in catalog_rows
}

catalog_exists = PROJECT_CATALOG in available_catalogs

print(f"Target catalog: {PROJECT_CATALOG}")
print(f"Already exists: {catalog_exists}")
```

```python
# Check system catalog availability
system_catalog_available = "system" in available_catalogs

print(
    "System catalog visible:",
    system_catalog_available
)

if not system_catalog_available:
    print(
        "WARNING: System catalog is not visible to this identity. "
        "Audit/system-table functionality will need to be verified later."
    )
```

```python
# Environment summary

environment_summary = {
    "project": PROJECT_NAME,
    "target_catalog": PROJECT_CATALOG,
    "spark_version": spark.version,
    "target_catalog_exists": catalog_exists,
    "system_catalog_visible": system_catalog_available
}

print("Environment validation summary")
print("=" * 60)

for key, value in environment_summary.items():
    print(f"{key}: {value}")
```

```python
# Fail only on actual blockers

blocking_issues = []

if spark is None:
    blocking_issues.append("Spark session unavailable")

if blocking_issues:
    raise RuntimeError(
        "Environment validation failed: "
        + "; ".join(blocking_issues)
    )

print("Environment validation completed successfully.")
print("Ready for project resource provisioning.")
```
