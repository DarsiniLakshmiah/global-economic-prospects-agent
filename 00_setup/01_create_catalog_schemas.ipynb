```python
# Configuration

PROJECT_CATALOG = "worldbank_ai"

SCHEMAS = [
    "bronze",
    "silver",
    "gold",
    "rag",
    "ai",
    "monitoring"
]

print(f"Target catalog: {PROJECT_CATALOG}")
print(f"Schemas to create: {SCHEMAS}")
```

```python
# Create the catalog

spark.sql(f"""
CREATE CATALOG IF NOT EXISTS {PROJECT_CATALOG}
COMMENT 'Unity Catalog for the Global Economic Prospects Intelligence Agent'
""")

print(f"Catalog '{PROJECT_CATALOG}' is ready.")
```

```python
# Create the schemas


schema_comments = {
    "bronze": "Raw ingested World Bank API data and document metadata",
    "silver": "Cleaned, validated, and standardized World Bank data",
    "gold": "Analytics-ready macroeconomic datasets for SQL and AI tools",
    "rag": "Processed document chunks and RAG-related objects",
    "ai": "AI evaluation, agent, and model-related project objects",
    "monitoring": "Application, inference, and quality monitoring data"
}

for schema_name in SCHEMAS:

    comment = schema_comments[schema_name]

    spark.sql(f"""
        CREATE SCHEMA IF NOT EXISTS {PROJECT_CATALOG}.{schema_name}
        COMMENT '{comment}'
    """)

    print(f"Created/verified: {PROJECT_CATALOG}.{schema_name}")
```

```python
display(
    spark.sql(f"SHOW SCHEMAS IN {PROJECT_CATALOG}")
)
```

```python
# Validate each one programmatically


schema_rows = spark.sql(
    f"SHOW SCHEMAS IN {PROJECT_CATALOG}"
).collect()

available_schemas = {
    row["databaseName"]
    for row in schema_rows
}

missing_schemas = [
    schema
    for schema in SCHEMAS
    if schema not in available_schemas
]

if missing_schemas:
    raise RuntimeError(
        f"Missing expected schemas: {missing_schemas}"
    )

print("All required project schemas are available.")
```
